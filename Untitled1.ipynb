```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_csv('Titanic-Dataset.csv')

print("Dataset Shape:", df.shape)
print(df.head())

print(df.info())
```

```python
# 1. Check missing values
print(df.isnull().sum())

# 2. Impute missing 'Age' values with median age
df['Age'] = df['Age'].fillna(df['Age'].median())

# 3. Impute missing 'Embarked' with the most frequent value (mode)
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# 4. Drop 'Cabin' column (too many missing values)
df = df.drop(columns=['Cabin'], errors='ignore')

# 5. Remove duplicate rows if any exist
df = df.drop_duplicates()

# Verify cleaning
print(df.isnull().sum())

plt.figure(figsize=(6, 4))
sns.countplot(x='Survived', hue='Sex', data=df, palette='Set2')
plt.title('Survival Count by Gender (0 = Died, 1 = Survived)')
plt.xlabel('Survival Status')
plt.ylabel('Passenger Count')
plt.savefig('chart1_survival_gender.png', bbox_inches='tight')
plt.show()

plt.figure(figsize=(7, 5))
sns.boxplot(x='Pclass', y='Age', data=df, palette='Pastel1')
plt.title('Age Distribution Across Ticket Classes')
plt.xlabel('Ticket Class (1st, 2nd, 3rd)')
plt.ylabel('Age')
plt.savefig('chart2_age_pclass.png', bbox_inches='tight')
plt.show()

plt.figure(figsize=(7, 5))
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Numeric Feature Correlation Heatmap')
plt.savefig('chart3_correlation.png', bbox_inches='tight')
plt.show()
```
