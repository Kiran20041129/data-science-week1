```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
df = pd.read_csv('Titanic-Dataset.csv')

# Clean missing values briefly
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

plt.figure(figsize=(8, 5))
sns.barplot(x='Pclass', y='Survived', hue='Sex', data=df, ci=None, palette='viridis')
plt.title('Survival Rate across Ticket Class and Gender')
plt.ylabel('Survival Probability')
plt.xlabel('Ticket Class (1 = 1st, 2 = 2nd, 3 = 3rd)')
plt.savefig('week2_chart1_survival_class_sex.png', bbox_inches='tight')
plt.show()

plt.figure(figsize=(8, 5))
sns.kdeplot(data=df, x='Age', hue='Survived', common_norm=False, fill=True, palette='crest', alpha=0.5)
plt.title('Age Density Distribution of Survivors vs Non-Survivors')
plt.xlabel('Age')
plt.ylabel('Density')
plt.savefig('week2_chart2_age_kde.png', bbox_inches='tight')
plt.show()

plt.figure(figsize=(8, 5))
sns.violinplot(x='Pclass', y='Fare', data=df[df['Fare'] < 200], palette='magma')
plt.title('Fare Range Distribution Across Passenger Classes')
plt.xlabel('Ticket Class')
plt.ylabel('Fare ($)')
plt.savefig('week2_chart3_fare_violin.png', bbox_inches='tight')
plt.show()

plt.figure(figsize=(8, 5))
sns.countplot(x='Embarked', hue='Survived', data=df, palette='Set2')
plt.title('Survival Count Based on Embarkation Port (S, C, Q)')
plt.xlabel('Port of Embarkation')
plt.ylabel('Passenger Count')
plt.savefig('week2_chart4_embarked_survival.png', bbox_inches='tight')
plt.show()

plt.figure(figsize=(8, 6))
numeric_df = df[['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']]
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Heatmap of Numerical Feature Relationships')
plt.savefig('week2_chart5_heatmap.png', bbox_inches='tight')
plt.show()
```
